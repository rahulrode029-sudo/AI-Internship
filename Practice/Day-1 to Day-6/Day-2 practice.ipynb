```python
# conditional statments

marks = int(input("enter your marks :"))

if marks >= 101:
    print("marks is invalid :")

elif marks >=90:
    print("grade : A")
    
elif marks >= 75 :
    print("grade : B")
    
elif marks >= 45 :
    print("grade : c")
    
else :
    print("fail")
```
